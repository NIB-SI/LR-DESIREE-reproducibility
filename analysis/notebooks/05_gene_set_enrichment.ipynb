## Gene set enrichment of DIU

```python
DIU_results = pd.read_csv("LR_DESIREE_PAPER/ANALYSIS/final_tables_paper/DIU_results.csv", index_col=0)
```

```python
DIU_results["Unitato_gene_id"] = DIU_results["gene_id"].str.rsplit(".", n=1).str[0]
# get genes that have DIU
DIU_genes = DIU_results[(DIU_results["FDR"] < 0.05) & (DIU_results["ratio_difference"]> 0.1)]["Unitato_gene_id"].unique().tolist()
```

```python
all_genes = DIU_results['Unitato_gene_id'].unique().tolist()
```

```python
# remove genes that contain bambu or Xloc or agat
DIU_genes = [gene for gene in DIU_genes if not any(x in gene for x in ["Bambu", "LOC", "agat"])] 
```

```python
len(DIU_genes)
```

```python
import matplotlib.pyplot as plt


# Perform enrichment
enr = gp.enrichr(
    gene_list=DIU_genes, 
    gene_sets='reference/ADAPT_liftoff/UniTato_Mercator4_gene-level.gmt',
    background = all_genes
)
    
# Clean up term names
    #enr.res2d.Term = enr.res2d.Term.str.split(" \(GO").str[0]
    
# Plot
gp.dotplot(enr.res2d, figsize=(2,5), title=f"DIU", cmap=plt.cm.autumn_r)
plt.show()
```

```python
enr.res2d.sort_values("Combined Score", ascending=False)["Genes"].iloc[0]
```

```python
second_enr_genes = enr.res2d.sort_values("P-value", ascending=True)["Genes"].iloc[1]
# split the genes by ; and remove any whitespace
second_enr_genes = second_enr_genes.split(";")
second_enr_genes = [gene.strip() for gene in second_enr_genes]
```

```python
DIU_results[DIU_results['Unitato_gene_id'].isin(second_enr_genes)][["Unitato_gene_id", "functional_annotation"]].drop_duplicates().sort_values("Unitato_gene_id")
```

## Enrichment of genes with allelic change in heat

```python
trans_results = "LR_DESIREE_PAPER/ANALYSIS/final_tables_paper/trans_allele_pairwise_with_expression_change.csv"
trans_results = pd.read_csv(trans_results, index_col=0)
```

```python
trans_results["Unitato_gene_id"] = trans_results["gene_id_i"].str.rsplit(".", n=1).str[0]
# get genes that have DIU
trans_genes = trans_results[(trans_results["FDR"] < 0.1) & (np.abs(trans_results["ratio_difference"]) > 0.1)]["Unitato_gene_id"].unique().tolist()
all_trans_genes = trans_results['Unitato_gene_id'].unique().tolist()
```

```python
import matplotlib.pyplot as plt


# Perform enrichment
enr = gp.enrichr(
    gene_list=trans_genes, 
    gene_sets='reference/ADAPT_liftoff/UniTato_Mercator4_gene-level.gmt',
    background = all_trans_genes
)
    
# Clean up term names
    #enr.res2d.Term = enr.res2d.Term.str.split(" \(GO").str[0]
    
# Plot
gp.dotplot(enr.res2d, figsize=(3,5), title=f"DIU", cmap=plt.cm.autumn_r, cutoff=0.05)
plt.show()
```
